```python
import os
import pandas as pd
import sqlalchemy
from pathlib import Path
from dotenv import load_dotenv
```

```python
load_dotenv()

host = os.getenv("SQL_HOST")
usr = os.getenv("SQL_USER")
pwd = os.getenv("SQL_PWD")
port = os.getenv("SQL_PORT")
db = os.getenv("SQL_DB")
```

```python


url = f"mysql+pymysql://{usr}:{pwd}@{host}:{port}/{db}"
con = sqlalchemy.create_engine(url)
```

```python
df = pd.read_csv("../data/olist_customers_dataset.csv")
```

```python
df.head()
```

```python
df.to_sql("olist_customers_dataset",con)
```

```python
p = Path("../data/")
for x in p.iterdir():
    name = Path(x).name
    tmp = pd.read_csv(x)
    try:
        tmp.to_sql(name.split(".")[0],con)
    except Exception as e:
        print(f"Error inserting data : {e}")
```
